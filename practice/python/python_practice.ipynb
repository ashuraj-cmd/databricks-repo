```python
employees = [
    {"name": "Amit", "department": "IT", "salary": 60000},
    {"name": "Rahul", "department": "HR", "salary": 50000},
    {"name": "Priya", "department": "IT", "salary": 75000},
    {"name": "Neha", "department": "Finance", "salary": 65000},
]
```

```python
#Write the Python code in your Databricks notebook.
it_employees = {emp["name"]: emp["salary"] for emp in employees if emp["department"] == "IT"}

print(it_employees)
```

```python
#create a dictionary containing IT employees whose salary is greater than 60,000.
it_employees = {emp["name"]: emp["salary"] for emp in employees if emp["department"] == "IT" and emp["salary"] > 60000}
print(it_employees)
```
